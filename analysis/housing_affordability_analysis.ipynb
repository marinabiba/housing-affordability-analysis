```python
!git clone https://github.com/marinabiba/housing-affordability-analysis.git

%cd housing-affordability-analysis

import pandas as pd

zhvi = pd.read_csv("data/Metro_zhvi_uc_sfrcondo_tier_0.33_0.67_sm_sa_month.csv")
zori = pd.read_csv("data/Metro_zori_uc_sfrcondomfr_sm_month.csv")
mortgage = pd.read_csv("data/MORTGAGE30US.csv")

print("Home Value Data:")
display(zhvi.head())

print("Rent Data:")
display(zori.head())

print("Mortgage Rate Data:")
display(mortgage.head())
```

```python
# Select St. Louis, Missouri
city = "St. Louis, MO"

zhvi_city = zhvi[zhvi["RegionName"] == city].copy()
zori_city = zori[zori["RegionName"] == city].copy()

print("St. Louis Home Value Data:")
display(zhvi_city)

print("St. Louis Rent Data:")
display(zori_city)
```

```python
# Compare St. Louis housing values and rent: May 2021 vs. May 2026

home_2021 = zhvi_city["2021-05-31"].iloc[0]
home_2026 = zhvi_city["2026-05-31"].iloc[0]

rent_2021 = zori_city["2021-05-31"].iloc[0]
rent_2026 = zori_city["2026-05-31"].iloc[0]

home_change = ((home_2026 - home_2021) / home_2021) * 100
rent_change = ((rent_2026 - rent_2021) / rent_2021) * 100

print("ST. LOUIS HOUSING AFFORDABILITY")
print("--------------------------------")

print(f"Home Value - May 2021: ${home_2021:,.2f}")
print(f"Home Value - May 2026: ${home_2026:,.2f}")
print(f"Home Value Increase: {home_change:.1f}%")

print()

print(f"Rent - May 2021: ${rent_2021:,.2f}")
print(f"Rent - May 2026: ${rent_2026:,.2f}")
print(f"Rent Increase: {rent_change:.1f}%")
```

```python
# Prepare mortgage rate data

mortgage["observation_date"] = pd.to_datetime(mortgage["observation_date"])

rate_2021 = mortgage[
    (mortgage["observation_date"] >= "2021-05-01") &
    (mortgage["observation_date"] <= "2021-05-31")
]["MORTGAGE30US"].mean()

rate_2026 = mortgage[
    (mortgage["observation_date"] >= "2026-05-01") &
    (mortgage["observation_date"] <= "2026-05-31")
]["MORTGAGE30US"].mean()

print(f"Average 30-Year Mortgage Rate - May 2021: {rate_2021:.2f}%")
print(f"Average 30-Year Mortgage Rate - May 2026: {rate_2026:.2f}%")
```

```python
# Estimate monthly mortgage payments
# Assumption: 20% down payment, 30-year fixed mortgage

down_payment_pct = 0.20
loan_term_years = 30
number_of_payments = loan_term_years * 12

loan_2021 = home_2021 * (1 - down_payment_pct)
loan_2026 = home_2026 * (1 - down_payment_pct)

monthly_rate_2021 = (rate_2021 / 100) / 12
monthly_rate_2026 = (rate_2026 / 100) / 12

payment_2021 = loan_2021 * (
    monthly_rate_2021 * (1 + monthly_rate_2021) ** number_of_payments
) / (
    (1 + monthly_rate_2021) ** number_of_payments - 1
)

payment_2026 = loan_2026 * (
    monthly_rate_2026 * (1 + monthly_rate_2026) ** number_of_payments
) / (
    (1 + monthly_rate_2026) ** number_of_payments - 1
)

mortgage_change = ((payment_2026 - payment_2021) / payment_2021) * 100

print("ESTIMATED MONTHLY MORTGAGE PAYMENT")
print("----------------------------------")
print(f"May 2021: ${payment_2021:,.2f}")
print(f"May 2026: ${payment_2026:,.2f}")
print(f"Increase: {mortgage_change:.1f}%")
```

```python
import matplotlib.pyplot as plt

categories = ["May 2021", "May 2026"]
rent_values = [rent_2021, rent_2026]
mortgage_values = [payment_2021, payment_2026]

x = range(len(categories))
width = 0.35

plt.figure(figsize=(9, 6))

plt.bar([i - width/2 for i in x], rent_values, width, label="Monthly Rent")
plt.bar([i + width/2 for i in x], mortgage_values, width, label="Mortgage Payment")

plt.xticks(x, categories)
plt.ylabel("Monthly Cost ($)")
plt.title("St. Louis Monthly Housing Cost: Rent vs. Mortgage")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Compare percentage increases from 2021 to 2026

categories = ["Home Value", "Monthly Rent", "Mortgage Payment"]
percent_changes = [home_change, rent_change, mortgage_change]

plt.figure(figsize=(9, 6))
bars = plt.bar(categories, percent_changes)

plt.ylabel("Increase (%)")
plt.title("Change in St. Louis Housing Costs: May 2021 to May 2026")

# Add percentage labels
for bar, value in zip(bars, percent_changes):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 2,
        f"{value:.1f}%",
        ha="center"
    )

plt.tight_layout()
plt.show()
```

```python
# Convert St. Louis housing data from wide to long format

id_columns = ["RegionID", "SizeRank", "RegionName", "RegionType", "StateName"]

home_trend = zhvi_city.melt(
    id_vars=id_columns,
    var_name="Date",
    value_name="Home Value"
)

rent_trend = zori_city.melt(
    id_vars=id_columns,
    var_name="Date",
    value_name="Rent"
)

home_trend["Date"] = pd.to_datetime(home_trend["Date"])
rent_trend["Date"] = pd.to_datetime(rent_trend["Date"])

# Keep May 2021 through May 2026
home_trend = home_trend[
    (home_trend["Date"] >= "2021-05-01") &
    (home_trend["Date"] <= "2026-05-31")
]

rent_trend = rent_trend[
    (rent_trend["Date"] >= "2021-05-01") &
    (rent_trend["Date"] <= "2026-05-31")
]

print("Home observations:", len(home_trend))
print("Rent observations:", len(rent_trend))
```

```python
fig, ax1 = plt.subplots(figsize=(10, 6))

ax1.plot(home_trend["Date"], home_trend["Home Value"], label="Home Value")
ax1.set_xlabel("Date")
ax1.set_ylabel("Home Value ($)")
ax1.set_title("St. Louis Housing Trends: May 2021 to May 2026")

ax2 = ax1.twinx()
ax2.plot(rent_trend["Date"], rent_trend["Rent"], label="Monthly Rent")
ax2.set_ylabel("Monthly Rent ($)")

fig.tight_layout()
plt.show()
```

```python
rent_vs_mortgage_2021 = rent_2021 - payment_2021
rent_vs_mortgage_2026 = rent_2026 - payment_2026

print("FINAL AFFORDABILITY SUMMARY")
print("---------------------------")
print(f"May 2021 rent exceeded estimated mortgage payment by ${rent_vs_mortgage_2021:,.2f} per month.")
print(f"May 2026 rent exceeded estimated mortgage payment by only ${rent_vs_mortgage_2026:,.2f} per month.")
print()
print("Conclusion:")
print(
    "Between May 2021 and May 2026, St. Louis home values increased 26.4% and rent increased 32.5%. "
    "However, the estimated monthly mortgage payment increased 89.3% because mortgage rates rose from "
    "2.96% to 6.44%. Buying still had a slightly lower estimated principal-and-interest payment in May 2026, "
    "but the monthly advantage over renting narrowed substantially."
)
```
